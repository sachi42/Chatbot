```python
import os
from dotenv import load_dotenv

groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq
model = ChatGroq(model="llama-3.3-70b-versatile", groq_api_key=groq_api_key)
model
```

```python
from langchain_core.messages import HumanMessage
model.invoke([HumanMessage(content="Hello, my name is Sachi and I am a Senior Software Engineer")])
```

```python
from langchain_core.messages import AIMessage
model.invoke([
    HumanMessage(content="Hello, my name is Sachi and I am a Senior Software Engineer"),
    AIMessage(content="Hello Sachi! It's great to meet you. As a Senior Software Engineer, you must have a wealth of experience in software development. What kind of projects do you typically work on, and what technologies do you enjoy using the most?"),
    HumanMessage(content="What's my name and what do you know about me?")
    ])
```

```python
## Chat mesaage history
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store = {}
def session_message_history(session_id: str) -> ChatMessageHistory:
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]

with_msg_history = RunnableWithMessageHistory(model, session_message_history)
```

```python
config = { "configurable": {"session_id": "my-session1"}}
```

```python
res = with_msg_history.invoke([ HumanMessage(content="Hello, my name is Sachi and I am a Senior Software Engineer")], config=config)
res.content
```

```python
res = with_msg_history.invoke([HumanMessage(content="What's my name and what do you know about me?")], config=config)
res.content
```

```python
config1 = { "configurable": {"session_id": "my-session2"}}
res = with_msg_history.invoke([HumanMessage(content="What's my name and what do you know about me?")], config=config1)
res.content
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant. Answer the questions to the best of your ability."),
    MessagesPlaceholder(variable_name="messages"),
])

chain=prompt|model
```

```python
with_msg_history = RunnableWithMessageHistory(chain, session_message_history)
config = { "configurable": {"session_id": "my-session3"}}
res = with_msg_history.invoke([ HumanMessage(content="Hello, my name is Sachi and I am a Senior Software Engineer")], config=config)
res.content
```

```python
res= with_msg_history.invoke([HumanMessage(content="What's my name and what do you know about me?")], config=config)
res.content
```

```python
## Add more Input variables to the prompt template
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant. Answer the questions to the best of your ability in {language}."),
    MessagesPlaceholder(variable_name="messages"),
])
chain=prompt|model

chain.invoke({"messages":[HumanMessage(content="Hi My name is Sachi")],"language":"Hindi"})
```

```python
with_msg_history_chain = RunnableWithMessageHistory(
    chain,
    session_message_history,
    input_messages_key="messages",
    history_messages_key="messages",
)
config = { "configurable": {"session_id": "my-session4"}}
res = with_msg_history_chain.invoke({"messages":[HumanMessage(content="Hi My name is Sachi")],"language":"Hindi"}, config=config)
res.content
```

```python
## Adding more complexity by using trim_messages to trim the messages in the history and using SystemMessage
from langchain_core.messages import SystemMessage, trim_messages
trimmer=trim_messages(
    max_tokens=45,
    strategy="last",
    token_counter=model,
    include_system=True,
    allow_partial=False,
    start_on="human"
)
messages = [
    SystemMessage(content="you're a good assistant"),
    HumanMessage(content="hi! I'm bob"),
    AIMessage(content="hi!"),
    HumanMessage(content="I like vanilla ice cream"),
    AIMessage(content="nice"),
    HumanMessage(content="whats 2 + 2"),
    AIMessage(content="4"),
    HumanMessage(content="thanks"),
    AIMessage(content="no problem!"),
    HumanMessage(content="having fun?"),
    AIMessage(content="yes!"),
]
trimmer.invoke(messages)
```

```python
## creating a chain with the trimmer and the model
from operator import itemgetter
from langchain_core.runnables import RunnablePassthrough

chain=(
    RunnablePassthrough.assign(messages=itemgetter("messages")|trimmer)
    | prompt
    | model
    
)
response=chain.invoke(
    {
    "messages":messages + [HumanMessage(content="What ice cream do i like")],
    "language":"English"
    }
)
response.content
```

```python
response=chain.invoke(
    {
    "messages":messages + [HumanMessage(content="What math problem did I ask you?")],
    "language":"English"
    }
)
response.content
```

```python
## Wrapping in a message history chain
with_msg_history_chain = RunnableWithMessageHistory(
    chain,
    session_message_history,
    input_messages_key="messages"
)
config = { "configurable": {"session_id": "my-session5"}}
res = with_msg_history_chain.invoke({"messages":messages + [HumanMessage(content="What math problem did I ask you?")],"language":"English"}, config=config)
res.content
```
